```python
# Temperature Converter: Celsius <-> Fahrenheit

def celsius_to_fahrenheit(celsius):
    """Convert Celsius to Fahrenheit."""
    return (celsius * 9/5) + 32

def fahrenheit_to_celsius(fahrenheit):
    """Convert Fahrenheit to Celsius."""
    return (fahrenheit - 32) * 5/9

def main():
    print("Temperature Converter")
    print("1. Celsius to Fahrenheit")
    print("2. Fahrenheit to Celsius")

    # Get user's choice of conversion
    choice = input("Choose an option (1 or 2): ")

    if choice == "1":
        celsius = float(input("Enter temperature in Celsius: "))
        fahrenheit = celsius_to_fahrenheit(celsius)
        print(f"{celsius}°C is equal to {fahrenheit:.2f}°F")

    elif choice == "2":
        fahrenheit = float(input("Enter temperature in Fahrenheit: "))
        celsius = fahrenheit_to_celsius(fahrenheit)
        print(f"{fahrenheit}°F is equal to {celsius:.2f}°C")

    else:
        print("Invalid choice. Please enter 1 or 2.")

if __name__ == "__main__":
    main()
```

```python
# Menu-Driven Calculator

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    # Handle division by zero
    if b == 0:
        return "Error: Cannot divide by zero"
    return a / b

def main():
    while True:  # Loop keeps the calculator running until user exits
        print("\n----- Menu-Driven Calculator -----")
        print("1. Addition (+)")
        print("2. Subtraction (-)")
        print("3. Multiplication (×)")
        print("4. Division (÷)")
        print("5. Exit")

        choice = input("Choose an operation (1-5): ")

        # Exit option handled first so we don't ask for numbers unnecessarily
        if choice == "5":
            print("Exiting calculator. Goodbye!")
            break

        # Validate menu choice before asking for numbers
        if choice not in ("1", "2", "3", "4"):
            print("Invalid choice. Please select a number between 1 and 5.")
            continue

        # Get the two numbers to operate on
        try:
            num1 = float(input("Enter the first number: "))
            num2 = float(input("Enter the second number: "))
        except ValueError:
            print("Invalid input. Please enter numeric values.")
            continue

        # Perform the chosen operation
        if choice == "1":
            result = add(num1, num2)
            print(f"Result: {num1} + {num2} = {result}")
        elif choice == "2":
            result = subtract(num1, num2)
            print(f"Result: {num1} - {num2} = {result}")
        elif choice == "3":
            result = multiply(num1, num2)
            print(f"Result: {num1} × {num2} = {result}")
        elif choice == "4":
            result = divide(num1, num2)
            print(f"Result: {num1} ÷ {num2} = {result}")

if __name__ == "__main__":
    main()
```

```python
"""
Student Gradebook Application
------------------------------
A simple console-based program that stores and manages student marks
using a CSV file. Supports adding students, viewing all records,
searching by name, and calculating grades.
"""

import csv
import os

# Name of the CSV file used to store student records
FILENAME = "students.csv"

# Column headers for the CSV file
FIELDNAMES = ["Name", "Marks", "Grade"]


def calculate_grade(marks):
    """
    Calculate a letter grade based on numeric marks (out of 100).
    Raises ValueError if marks are out of the valid 0-100 range.
    """
    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    elif marks >= 50:
        return "E"
    else:
        return "F"


def initialize_file():
    """
    Create the CSV file with headers if it does not already exist.
    """
    if not os.path.exists(FILENAME):
        try:
            with open(FILENAME, mode="w", newline="") as file:
                writer = csv.DictWriter(file, fieldnames=FIELDNAMES)
                writer.writeheader()
        except IOError as e:
            print(f"Error creating file: {e}")


def add_student():
    """
    Prompt the user for a student's name and marks, calculate the grade,
    and append the record to the CSV file.
    """
    try:
        name = input("Enter student name: ").strip()
        if not name:
            print("Error: Name cannot be empty.")
            return

        marks_input = input("Enter student marks (0-100): ").strip()
        marks = float(marks_input)  # may raise ValueError if not numeric

        grade = calculate_grade(marks)  # may raise ValueError if out of range

        with open(FILENAME, mode="a", newline="") as file:
            writer = csv.DictWriter(file, fieldnames=FIELDNAMES)
            writer.writerow({"Name": name, "Marks": marks, "Grade": grade})

        print(f"Student '{name}' added successfully with grade '{grade}'.")

    except ValueError as e:
        # Handles both invalid numeric input and out-of-range marks
        print(f"Invalid input: {e}")
    except IOError as e:
        print(f"File error while saving student: {e}")


def view_students():
    """
    Read and display all student records from the CSV file.
    Handles the case where the file is missing or empty.
    """
    try:
        with open(FILENAME, mode="r", newline="") as file:
            reader = csv.DictReader(file)
            records = list(reader)

            if not records:
                print("No student records found.")
                return

            print("\n{:<20} {:<10} {:<10}".format("Name", "Marks", "Grade"))
            print("-" * 40)
            for row in records:
                print("{:<20} {:<10} {:<10}".format(
                    row["Name"], row["Marks"], row["Grade"]))

    except FileNotFoundError:
        print("Error: No records file found yet. Please add a student first.")
    except IOError as e:
        print(f"Error reading file: {e}")


def search_student():
    """
    Search for a student by name (case-insensitive) and display their
    marks and grade if found.
    """
    try:
        search_name = input("Enter student name to search: ").strip().lower()

        with open(FILENAME, mode="r", newline="") as file:
            reader = csv.DictReader(file)
            found = False

            for row in reader:
                if row["Name"].strip().lower() == search_name:
                    print(f"\nName: {row['Name']}")
                    print(f"Marks: {row['Marks']}")
                    print(f"Grade: {row['Grade']}")
                    found = True
                    break

            if not found:
                print(f"No record found for student '{search_name}'.")

    except FileNotFoundError:
        print("Error: No records file found yet. Please add a student first.")
    except IOError as e:
        print(f"Error reading file: {e}")


def display_menu():
    """Display the main menu options to the user."""
    print("\n===== Student Gradebook Menu =====")
    print("1. Add Student")
    print("2. View Student Records")
    print("3. Search Student")
    print("4. Exit")


def main():
    """Main program loop that drives the gradebook application."""
    initialize_file()

    while True:
        display_menu()
        choice = input("Enter your choice (1-4): ").strip()

        if choice == "1":
            add_student()
        elif choice == "2":
            view_students()
        elif choice == "3":
            search_student()
        elif choice == "4":
            print("Exiting Gradebook Application. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 4.")


# Entry point of the program
if __name__ == "__main__":
    main()
```

```python
"""
Currency Converter using the exchangerate-api.com free API

Week 8 Assignment: Build a Currency Converter Using an Exchange Rate API
"""

import requests

# Free, no-API-key-required endpoint (open.er-api.com).
# It returns the latest exchange rates for a given base currency.
BASE_URL = "https://open.er-api.com/v6/latest/"


def get_exchange_rate(source_currency, target_currency):
    """
    Fetch the exchange rate from source_currency to target_currency.

    Returns the rate as a float, or None if something went wrong
    (invalid currency code, network error, bad response, etc.)
    """
    url = BASE_URL + source_currency.upper()

    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()  # raises an error for 4xx/5xx status codes
    except requests.exceptions.RequestException as e:
        print(f"Error: Could not connect to the exchange rate API. ({e})")
        return None

    # Parse the JSON body of the response
    data = response.json()

    # The API returns "result": "error" if the base currency code is invalid
    if data.get("result") != "success":
        print("Error: Invalid source currency code or API returned an error.")
        return None

    rates = data.get("rates", {})
    target_currency = target_currency.upper()

    if target_currency not in rates:
        print("Error: Invalid target currency code.")
        return None

    return rates[target_currency]


def get_user_input():
    """
    Prompt the user for the source currency, target currency, and amount.
    Returns a tuple (source, target, amount) or None if the amount is invalid.
    """
    source_currency = input("Enter source currency (e.g., USD): ").strip().upper()
    target_currency = input("Enter target currency (e.g., EUR): ").strip().upper()

    try:
        amount = float(input("Enter the amount to convert: ").strip())
    except ValueError:
        print("Error: Amount must be a number.")
        return None

    if amount < 0:
        print("Error: Amount cannot be negative.")
        return None

    return source_currency, target_currency, amount


def convert_currency(amount, rate):
    """Calculate the converted amount given the amount and exchange rate."""
    return amount * rate


def main():
    """Main program loop: get input, fetch rate, convert, and display result."""
    user_data = get_user_input()

    # Stop early if the amount input was invalid
    if user_data is None:
        return

    source_currency, target_currency, amount = user_data

    rate = get_exchange_rate(source_currency, target_currency)

    # Stop early if the exchange rate could not be retrieved
    if rate is None:
        return

    converted_amount = convert_currency(amount, rate)

    print("\n--- Conversion Result ---")
    print(f"Exchange rate: 1 {source_currency} = {rate} {target_currency}")
    print(f"{amount} {source_currency} = {converted_amount:.2f} {target_currency}")


if __name__ == "__main__":
    main()
```

```python
class Student:
    """Base class representing a general student."""
    def __init__(self, student_id, name, age, course):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.course = course

    def display_info(self):
        """Displays student details."""
        return (f"ID: {self.student_id} | Name: {self.name} | "
                f"Age: {self.age} | Course: {self.course}")


class GraduateStudent(Student):
    """Subclass demonstrating inheritance and method overriding."""
    def __init__(self, student_id, name, age, course, thesis_topic):
        super().__init__(student_id, name, age, course)
        self.thesis_topic = thesis_topic

    # Method Overriding
    def display_info(self):
        """Overrides base class method to include thesis info."""
        base_info = super().display_info()
        return f"{base_info} | Thesis: {self.thesis_topic}"


class StudentManagementSystem:
    """Class to manage multiple students."""
    def __init__(self):
        self.students = {}

    def add_student(self):
        """Adds a new student to the system."""
        student_id = input("Enter Student ID: ")
        if student_id in self.students:
            print("Error: Student ID already exists.")
            return

        name = input("Enter Name: ")
        age = input("Enter Age: ")
        course = input("Enter Course/Program: ")
        is_grad = input("Is this a graduate student? (y/n): ").lower()

        if is_grad == 'y':
            thesis = input("Enter Thesis Topic: ")
            student = GraduateStudent(student_id, name, age, course, thesis)
        else:
            student = Student(student_id, name, age, course)

        self.students[student_id] = student
        print("Student added successfully!")

    def update_student(self):
        """Updates an existing student's information using Student ID."""
        student_id = input("Enter Student ID to update: ")
        if student_id not in self.students:
            print("Error: Student ID not found.")
            return

        print("Leave field blank if you do not want to change it.")
        name = input("Enter new Name: ")
        age = input("Enter new Age: ")
        course = input("Enter new Course/Program: ")

        student = self.students[student_id]
        if name:
            student.name = name
        if age:
            student.age = age
        if course:
            student.course = course

        print("Student updated successfully!")

    def view_students(self):
        """Displays all registered students."""
        if not self.students:
            print("No students registered yet.")
            return

        print("\n--- Registered Students ---")
        for student in self.students.values():
            print(student.display_info())
        print("-" * 27)


def main():
    system = StudentManagementSystem()
    while True:
        print("\n=== Student Management System ===")
        print("1. Add Student")
        print("2. Update Student")
        print("3. View Students")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            system.add_student()
        elif choice == '2':
            system.update_student()
        elif choice == '3':
            system.view_students()
        elif choice == '4':
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number from 1 to 4.")

if __name__ == "__main__":
    main()
```
